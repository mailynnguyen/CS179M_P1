```python
import matplotlib.pyplot as plt
import numpy as np

datasets = ["128Square4", "128Circle201", "256Cashew"]
initial = np.array([430, 320, 850])
ten_sec = np.array([380, 300, 770])
hundred_sec = np.array([360, 290, 730])

percent_10 = (ten_sec / initial) * 100
percent_100 = (hundred_sec / initial) * 100

plt.figure(figsize=(6, 4))
time_points = [0, 10, 100]
for i, name in enumerate(datasets):
    plt.plot(time_points, [100, percent_10[i], percent_100[i]], marker="o", label=name)
plt.xscale("log")
plt.xlabel("Runtime (seconds)")
plt.ylabel("Route Distance (% of Initial)")
plt.title("Figure 1. Anytime Convergence of ComputeDronePath")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig("Anytime_Convergence.png", dpi=300)

plt.figure(figsize=(6, 4))
improvement = ((initial - hundred_sec) / initial) * 100
plt.bar(datasets, improvement, color=["#4e79a7", "#f28e2b", "#76b7b2"])
plt.xlabel("Dataset")
plt.ylabel("Improvement (%)")
plt.title("Figure 2. Improvement Across Datasets")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.savefig("Improvement_Across_Datasets.png", dpi=300)

print("Graphs saved as Anytime_Convergence.png and Improvement_Across_Datasets.png")
```
